## Step 5: Test with a Simple Agent

Let's create the simplest possible Pydantic AI agent to verify everything works!

This agent will:
1. Use Claude Sonnet 4.0 (fast and affordable)
2. Answer a simple question
3. Show us the response

````python
from pydantic_ai import Agent

# Create a simple agent
agent = Agent(
    'anthropic:claude-sonnet-4-0',  # Model to use
    instructions='You are a helpful assistant. Be concise.',  # System prompt
)

# Run the agent synchronously
result = agent.run_sync('What is 2 + 2?')

print("Agent response:")
print(result.output)
print("\nToken usage:")
print(result.usage())
````

Let's break down what just happened...

## Understanding the Code

### Agent Creation

```python
agent = Agent(
    'anthropic:claude-sonnet-4-0',  # Model specification
    instructions='You are a helpful assistant.',  # System prompt
)
```

- **Model format**: `'provider:model-name'`
- **Instructions**: The system prompt that guides the agent's behavior

### Running the Agent

```python
result = agent.run_sync('What is 2 + 2?')
```

- `run_sync()`: Synchronous execution (blocks until response)
- Returns a `RunResult` object with the response and metadata

### Accessing Results

```python
result.output    # The agent's text response
result.usage()   # Token usage information
```

## Testing Different Models

Let's try a few different models to see how they respond:

````python
# Test multiple models (if you have API keys configured)
models_to_test = [
    'anthropic:claude-sonnet-4-0',
    'anthropic:claude-haiku-4-0',
]

question = "Explain Pydantic AI in one sentence."

for model in models_to_test:
    try:
        agent = Agent(model, instructions='Be concise.')
        result = agent.run_sync(question)
        print(f"\n{model}:")
        print(f"  {result.output}")
        print(f"  Tokens: {result.usage().total_tokens}")
    except Exception as e:
        print(f"\n{model}: ❌ Error - {e}")
````
